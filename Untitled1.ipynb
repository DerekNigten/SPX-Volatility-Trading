```python
"""
Calendar Spread Trading Rule v2
================================
Improvements over v1:
  - Dynamic aggressiveness based on PC1 regime (was fixed)
  - Minimum holding period to reduce unnecessary turnover
  - Smarter expiration: roll 3 days early to avoid gamma/pin risk

Signal logic:
  - LONG calendar  (sell 30d, buy 60d): slope_forecast negative + calm/moderate regime
  - SHORT calendar (buy 30d, sell 60d): slope_forecast positive + stressed regime
  - FLAT: low conviction or ambiguous regime
"""

import pandas as pd
import numpy as np
from dataclasses import dataclass


@dataclass
class RuleConfig:
    """
    All tunable parameters in one place.
    
    base_aggressiveness: int (1-10)
        Starting point. Actual aggressiveness is adjusted by PC1 regime.
    
    dynamic_agg: bool
        If True, aggressiveness varies with PC1 regime:
          calm (low PC1)     → base + 2  (signals very reliable)
          moderate            → base
          stressed (high PC1) → base - 2  (signals degrade)
    
    min_hold_days: int
        Minimum days to hold before allowing position change. Reduces churn.
    
    roll_days_before_expiry: int
        Days before expiration to roll positions. Avoids gamma/pin risk.
    
    expiry_cycle_days: int
        Trading days between expirations (30 = monthly cycle).
    """
    base_aggressiveness: int = 5
    dynamic_agg: bool = True
    min_hold_days: int = 3
    max_long: int = 3
    max_short: int = 2
    roll_days_before_expiry: int = 3
    expiry_cycle_days: int = 30


def _get_effective_aggressiveness(base_agg: int, pc1: float, dynamic: bool) -> int:
    """Adjust aggressiveness based on PC1 regime."""
    if not dynamic:
        return base_agg
    if pc1 < -0.5:
        adj = 2
    elif pc1 < 1.0:
        adj = 0
    elif pc1 < 2.0:
        adj = -2
    else:
        adj = -3
    return int(np.clip(base_agg + adj, 1, 10))


def generate_signal(row: pd.Series, config: RuleConfig) -> dict:
    """Core trading rule. Returns {'long_calendars': X, 'short_calendars': Y}."""
    forecast = row['slope_forecast']
    pc1 = row['pc1']
    agg = _get_effective_aggressiveness(config.base_aggressiveness, pc1, config.dynamic_agg)

    min_forecast = np.interp(agg, [1, 10], [0.40, 0.05])
    pc1_max_long = np.interp(agg, [1, 10], [0.5, 2.5])
    pc1_min_short = np.interp(agg, [1, 10], [2.0, 0.8])
    min_forecast_short = np.interp(agg, [1, 10], [0.25, 0.05])

    size_thresholds = [
        np.interp(agg, [1, 10], [0.30, 0.10]),
        np.interp(agg, [1, 10], [0.50, 0.30]),
        np.interp(agg, [1, 10], [0.70, 0.50]),
    ]
    pc1_stress_penalty_threshold = np.interp(agg, [1, 10], [0.3, 1.5])

    abs_forecast = abs(forecast)
    long_size = 0
    short_size = 0

    if forecast < 0 and abs_forecast >= min_forecast and pc1 < pc1_max_long:
        base_size = 0
        for i, thresh in enumerate(size_thresholds):
            if abs_forecast >= thresh:
                base_size = i + 1
        if pc1 > pc1_stress_penalty_threshold:
            base_size = max(base_size - 1, 0)
        long_size = min(base_size, config.max_long)

    elif forecast > 0 and abs_forecast >= min_forecast_short and pc1 > pc1_min_short:
        base_size = 0
        for i, thresh in enumerate(size_thresholds):
            if abs_forecast >= thresh:
                base_size = i + 1
        short_size = min(base_size, config.max_short)

    return {'long_calendars': long_size, 'short_calendars': short_size}


def apply_holding_period(signals_df: pd.DataFrame, min_hold: int) -> pd.DataFrame:
    """Enforce minimum holding period to reduce churn."""
    if min_hold <= 1:
        return signals_df
    
    result = signals_df.copy()
    held_long = 0
    held_short = 0
    days_in_position = 0
    filtered_signals = []
    
    for idx, row in result.iterrows():
        raw = row['signal']
        raw_long = raw['long_calendars']
        raw_short = raw['short_calendars']
        position_changes = (raw_long != held_long) or (raw_short != held_short)
        
        if position_changes and days_in_position < min_hold and days_in_position > 0:
            filtered_signals.append({'long_calendars': held_long, 'short_calendars': held_short})
            days_in_position += 1
        else:
            filtered_signals.append({'long_calendars': raw_long, 'short_calendars': raw_short})
            if position_changes:
                days_in_position = 1
            else:
                days_in_position += 1
            held_long = raw_long
            held_short = raw_short
    
    result['signal'] = filtered_signals
    return result
```

```python
"""
Calendar Spread Strategy Simulation v2
Improvements: dynamic aggressiveness, min holding period, smart expiration rolling
"""

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ============================================================
# STEP 1: LOAD AND PREPARE REAL DATA
# ============================================================

df_real = pd.read_parquet('calendar_spread_data.parquet')

# Load signal data and merge
df_signals = pd.read_csv('calendar_spread_signals.csv', parse_dates=['date'])
df_real = df_real.merge(df_signals[['date', 'slope_forecast', 'pc1']], on='date', how='inner')
df_real = df_real.reset_index(drop=True)

# ── Configure strategy ───────────────────────────────────────
config = RuleConfig(
    base_aggressiveness=5,
    dynamic_agg=True,           # NEW: PC1-adaptive aggressiveness
    min_hold_days=3,            # NEW: reduce churn
    roll_days_before_expiry=0,  # NEW: roll early to avoid gamma/pin risk
    expiry_cycle_days=30,
)

# Generate raw signals
df_real['signal'] = df_real.apply(
    lambda row: generate_signal(row, config) 
    if pd.notna(row.get('slope_forecast')) 
    else {'long_calendars': 0, 'short_calendars': 0},
    axis=1
)

# Apply holding period filter (improvement #2)
df_real = apply_holding_period(df_real, config.min_hold_days)

print("Real data loaded successfully!")
print(f"Total trading days: {len(df_real)}")
print(f"Date range: {df_real['date'].min()} to {df_real['date'].max()}")
print(f"Config: base_agg={config.base_aggressiveness}, dynamic={config.dynamic_agg}, "
      f"min_hold={config.min_hold_days}d, roll={config.roll_days_before_expiry}d early")
print("\nFirst 5 days:")
print(df_real[['date', 'spx_price', 'signal']].head())

# ============================================================
# STEP 2: SIMULATION ENVIRONMENT
# ============================================================

class CalendarSpreadSimulator:
    def __init__(self, initial_cash=100000):
        self.long_calendars = 0
        self.short_calendars = 0
        self.cash = initial_cash
        self.initial_cash = initial_cash
        self.daily_pnl = []
        self.daily_portfolio_value = []
        self.transaction_costs = 0
        self.trades = []
        
    def get_portfolio_value(self, long_cal_value, short_cal_value):
        position_value = (self.long_calendars * long_cal_value - 
                         self.short_calendars * short_cal_value)
        return self.cash + position_value
    
    def get_current_positions(self):
        return {
            'long_calendars': self.long_calendars,
            'short_calendars': self.short_calendars,
            'cash': self.cash
        }


def execute_trades(simulator, row, use_costs=True):
    """Execute trades to reach target position from signal"""
    target_long = row['signal']['long_calendars']
    target_short = row['signal']['short_calendars']
    
    current_long = simulator.long_calendars
    current_short = simulator.short_calendars
    
    long_trade = target_long - current_long
    short_trade = target_short - current_short
    
    if use_costs:
        long_cal_buy_price = row['option_60dte_ask'] - row['option_30dte_bid']
        long_cal_sell_price = row['option_60dte_bid'] - row['option_30dte_ask']
        short_cal_buy_price = row['option_30dte_ask'] - row['option_60dte_bid']
        short_cal_sell_price = row['option_30dte_bid'] - row['option_60dte_ask']
    else:
        long_cal_price = row['option_60dte_mid'] - row['option_30dte_mid']
        long_cal_buy_price = long_cal_sell_price = long_cal_price
        short_cal_price = row['option_30dte_mid'] - row['option_60dte_mid']
        short_cal_buy_price = short_cal_sell_price = short_cal_price
    
    total_cost = 0
    
    if long_trade > 0:
        cost = long_trade * long_cal_buy_price
        simulator.cash -= cost
        simulator.long_calendars += long_trade
        total_cost += abs(long_cal_buy_price - (row['option_60dte_mid'] - row['option_30dte_mid'])) * long_trade
        simulator.trades.append({
            'date': row['date'], 'type': 'BUY_LONG_CAL',
            'quantity': long_trade, 'price': long_cal_buy_price, 'cost': cost
        })
    elif long_trade < 0:
        proceeds = abs(long_trade) * long_cal_sell_price
        simulator.cash += proceeds
        simulator.long_calendars += long_trade
        total_cost += abs(long_cal_sell_price - (row['option_60dte_mid'] - row['option_30dte_mid'])) * abs(long_trade)
        simulator.trades.append({
            'date': row['date'], 'type': 'SELL_LONG_CAL',
            'quantity': abs(long_trade), 'price': long_cal_sell_price, 'proceeds': proceeds
        })
    
    if short_trade > 0:
        proceeds = short_trade * short_cal_sell_price
        simulator.cash += proceeds
        simulator.short_calendars += short_trade
        total_cost += abs(short_cal_sell_price - (row['option_30dte_mid'] - row['option_60dte_mid'])) * short_trade
        simulator.trades.append({
            'date': row['date'], 'type': 'OPEN_SHORT_CAL',
            'quantity': short_trade, 'price': short_cal_sell_price, 'proceeds': proceeds
        })
    elif short_trade < 0:
        cost = abs(short_trade) * short_cal_buy_price
        simulator.cash -= cost
        simulator.short_calendars += short_trade
        total_cost += abs(short_cal_buy_price - (row['option_30dte_mid'] - row['option_60dte_mid'])) * abs(short_trade)
        simulator.trades.append({
            'date': row['date'], 'type': 'CLOSE_SHORT_CAL',
            'quantity': abs(short_trade), 'price': short_cal_buy_price, 'cost': cost
        })
    
    simulator.transaction_costs += total_cost
    return total_cost


def handle_expiration(simulator, current_day_idx, df_data, config):
    """
    Improved expiration handling: roll positions BEFORE expiry.
    
    Instead of waiting until exact expiry (day % 30 == 0), we roll 
    `roll_days_before_expiry` days early. This avoids:
      - Gamma risk from the short 30 DTE leg near expiration
      - Pin risk if SPX is near the strike
      - Liquidity issues on expiration day
    
    The roll closes all positions at mid price. The simulation loop 
    will then re-enter based on the signal for that day.
    """
    cycle = config.expiry_cycle_days
    roll_early = config.roll_days_before_expiry
    
    # Roll day = cycle - roll_early days after cycle start
    # i.e., if cycle=30 and roll_early=3, roll on day 27 of each cycle
    roll_point = cycle - roll_early
    
    if roll_point <= 0:
        roll_point = cycle  # fallback: roll on expiry day
    
    # Check if this is a roll day
    days_into_cycle = (current_day_idx + 1) % cycle
    if days_into_cycle != roll_point:
        return 0
    
    row = df_data.iloc[current_day_idx]
    expiration_cost = 0
    
    # Close all positions at mid price (rolling, not expiry settlement)
    # This is more realistic: we close before the short leg expires
    
    if simulator.long_calendars > 0:
        # Close long calendar: sell the 60 DTE leg, buy back 30 DTE leg
        long_cal_mid = row['option_60dte_mid'] - row['option_30dte_mid']
        closing_proceeds = simulator.long_calendars * long_cal_mid
        simulator.cash += closing_proceeds
        
        simulator.trades.append({
            'date': row['date'],
            'type': 'ROLL_CLOSE_LONG_CAL',
            'quantity': simulator.long_calendars,
            'price': long_cal_mid,
            'proceeds': closing_proceeds
        })
        
        simulator.long_calendars = 0
    
    if simulator.short_calendars > 0:
        # Close short calendar: buy back 30 DTE leg, sell 60 DTE leg
        short_cal_mid = row['option_30dte_mid'] - row['option_60dte_mid']
        closing_cost = simulator.short_calendars * short_cal_mid
        simulator.cash -= closing_cost
        
        simulator.trades.append({
            'date': row['date'],
            'type': 'ROLL_CLOSE_SHORT_CAL',
            'quantity': simulator.short_calendars,
            'price': short_cal_mid,
            'cost': closing_cost
        })
        
        simulator.short_calendars = 0
    
    return expiration_cost


# ============================================================
# STEP 3: RUN SIMULATION
# ============================================================

print("\n" + "="*60)
print("RUNNING SIMULATION")
print("="*60)

simulator = CalendarSpreadSimulator(initial_cash=100000)

for idx, row in df_real.iterrows():
    # Handle expiration/rolling (improvement #3)
    exp_cost = handle_expiration(simulator, idx, df_real, config)
    
    # Execute trades
    txn_cost = execute_trades(simulator, row, use_costs=True)
    
    # Calculate portfolio value
    long_cal_mid = row['option_60dte_mid'] - row['option_30dte_mid']
    short_cal_mid = row['option_30dte_mid'] - row['option_60dte_mid']
    portfolio_value = simulator.get_portfolio_value(long_cal_mid, short_cal_mid)
    
    # Calculate daily P&L
    if idx == 0:
        daily_pnl = 0
    else:
        daily_pnl = portfolio_value - simulator.daily_portfolio_value[-1]
    
    simulator.daily_pnl.append(daily_pnl)
    simulator.daily_portfolio_value.append(portfolio_value)
    simulator.transaction_costs += exp_cost

print("\nSimulation complete!")
print(f"Final positions:")
print(f"  Long calendars: {simulator.long_calendars}")
print(f"  Short calendars: {simulator.short_calendars}")
print(f"  Cash: ${simulator.cash:,.2f}")
print(f"  Final portfolio value: ${simulator.daily_portfolio_value[-1]:,.2f}")
print(f"  Total return: ${simulator.daily_portfolio_value[-1] - simulator.initial_cash:,.2f}")
print(f"  Total transaction costs: ${simulator.transaction_costs:,.2f}")
print(f"\nTotal trades executed: {len(simulator.trades)}")

# Count roll events
roll_trades = [t for t in simulator.trades if 'ROLL' in t['type']]
print(f"Roll events: {len(roll_trades)}")

# ============================================================
# STEP 4: PERFORMANCE METRICS
# ============================================================

def calculate_performance_metrics(simulator):
    total_return_dollar = simulator.daily_portfolio_value[-1] - simulator.initial_cash
    total_return_pct = (simulator.daily_portfolio_value[-1] / simulator.initial_cash - 1) * 100
    
    daily_returns = np.diff(simulator.daily_portfolio_value) / simulator.daily_portfolio_value[:-1]
    mean_daily_return = np.mean(daily_returns)
    std_daily_return = np.std(daily_returns)
    
    sharpe_ratio = (mean_daily_return / std_daily_return) * np.sqrt(252) if std_daily_return > 0 else 0
    
    portfolio_values = np.array(simulator.daily_portfolio_value)
    running_max = np.maximum.accumulate(portfolio_values)
    drawdown = (portfolio_values - running_max) / running_max * 100
    max_drawdown = np.min(drawdown)
    
    num_trades = len(simulator.trades)
    positive_days = sum(1 for pnl in simulator.daily_pnl if pnl > 0)
    total_days = len(simulator.daily_pnl)
    win_rate = (positive_days / total_days * 100) if total_days > 0 else 0
    avg_daily_pnl = np.mean(simulator.daily_pnl)
    
    return {
        'total_return_dollar': total_return_dollar,
        'total_return_pct': total_return_pct,
        'sharpe_ratio': sharpe_ratio,
        'max_drawdown_pct': max_drawdown,
        'num_trades': num_trades,
        'win_rate_pct': win_rate,
        'avg_daily_pnl': avg_daily_pnl,
        'total_transaction_costs': simulator.transaction_costs,
        'final_cash': simulator.cash,
        'final_portfolio_value': simulator.daily_portfolio_value[-1]
    }


def print_performance_report(metrics):
    print("\n" + "="*60)
    print("PERFORMANCE METRICS")
    print("="*60)
    print(f"\nReturns:")
    print(f"  Total Return:           ${metrics['total_return_dollar']:,.2f}")
    print(f"  Total Return (%):       {metrics['total_return_pct']:.2f}%")
    print(f"  Avg Daily P&L:          ${metrics['avg_daily_pnl']:,.2f}")
    print(f"\nRisk Metrics:")
    print(f"  Sharpe Ratio:           {metrics['sharpe_ratio']:.3f}")
    print(f"  Max Drawdown:           {metrics['max_drawdown_pct']:.2f}%")
    print(f"\nTrading Activity:")
    print(f"  Total Trades:           {metrics['num_trades']}")
    print(f"  Win Rate:               {metrics['win_rate_pct']:.2f}%")
    print(f"  Transaction Costs:      ${metrics['total_transaction_costs']:,.2f}")
    print(f"\nFinal State:")
    print(f"  Cash:                   ${metrics['final_cash']:,.2f}")
    print(f"  Portfolio Value:        ${metrics['final_portfolio_value']:,.2f}")
    print("="*60)


metrics = calculate_performance_metrics(simulator)
print_performance_report(metrics)

# ============================================================
# STEP 5: VISUALIZATIONS
# ============================================================

def plot_clean_performance(simulator):
    plt.style.use('seaborn-v0_8-whitegrid')
    fig, axes = plt.subplots(2, 2, figsize=(14, 9))
    fig.suptitle('Calendar Spread Strategy v2 (Dynamic Agg + Holding Period + Smart Roll)',
                 fontsize=14, fontweight='bold', y=0.98)
    
    # 1. Cumulative Returns
    ax1 = axes[0, 0]
    returns = (np.array(simulator.daily_portfolio_value) / simulator.initial_cash - 1) * 100
    ax1.plot(returns, linewidth=2.5, color='#2E86AB')
    ax1.axhline(y=0, color='black', linestyle='--', linewidth=1, alpha=0.7)
    ax1.fill_between(range(len(returns)), returns, 0, alpha=0.25, color='#2E86AB')
    ax1.set_title('Cumulative Returns', fontsize=13, fontweight='bold', pad=12)
    ax1.set_ylabel('Return (%)')
    ax1.set_xlabel('Trading Days')
    ax1.grid(True, alpha=0.4)
    
    # 2. Drawdown
    ax2 = axes[0, 1]
    portfolio_values = np.array(simulator.daily_portfolio_value)
    running_max = np.maximum.accumulate(portfolio_values)
    drawdown = (portfolio_values - running_max) / running_max * 100
    ax2.fill_between(range(len(drawdown)), drawdown, 0, alpha=0.7, color='#E63946')
    ax2.plot(drawdown, linewidth=1.5, color='#8B0000', alpha=0.8)
    ax2.set_title('Drawdown', fontsize=13, fontweight='bold', pad=12)
    ax2.set_ylabel('Drawdown (%)')
    ax2.set_xlabel('Trading Days')
    ax2.grid(True, alpha=0.4)
    
    # 3. Monthly P&L
    ax3 = axes[1, 0]
    days_per_month = 21
    num_months = len(simulator.daily_pnl) // days_per_month
    monthly_returns = []
    for i in range(num_months):
        start_idx = i * days_per_month
        end_idx = min((i + 1) * days_per_month, len(simulator.daily_pnl))
        monthly_returns.append(sum(simulator.daily_pnl[start_idx:end_idx]))
    
    colors = ['#E63946' if r < 0 else '#06A77D' for r in monthly_returns]
    ax3.bar(range(1, len(monthly_returns) + 1), monthly_returns, color=colors, alpha=0.85, edgecolor='black', linewidth=1)
    ax3.axhline(y=0, color='black', linestyle='-', linewidth=1.5)
    ax3.set_title('Monthly P&L', fontsize=13, fontweight='bold', pad=12)
    ax3.set_ylabel('P&L ($)')
    ax3.set_xlabel('Month')
    ax3.grid(True, alpha=0.4, axis='y')
    
    # 4. Metrics Summary
    ax4 = axes[1, 1]
    ax4.axis('off')
    
    total_return = (simulator.daily_portfolio_value[-1] / simulator.initial_cash - 1) * 100
    daily_returns_arr = np.diff(simulator.daily_portfolio_value) / simulator.daily_portfolio_value[:-1]
    sharpe = (np.mean(daily_returns_arr) / np.std(daily_returns_arr)) * np.sqrt(252) if np.std(daily_returns_arr) > 0 else 0
    max_dd = np.min(drawdown)
    win_rate = sum(1 for pnl in simulator.daily_pnl if pnl > 0) / len(simulator.daily_pnl) * 100
    roll_count = len([t for t in simulator.trades if 'ROLL' in t['type']])
    
    metrics_text = f"""
    PERFORMANCE SUMMARY (v2)
    {'─' * 35}
    
    Total Return:          {total_return:>10.2f}%
    
    Sharpe Ratio:          {sharpe:>10.2f}
    
    Max Drawdown:          {max_dd:>10.2f}%
    
    Win Rate:              {win_rate:>10.1f}%
    
    Total Trades:          {len(simulator.trades):>10}
    
    Roll Events:           {roll_count:>10}
    
    Transaction Costs:     ${simulator.transaction_costs:>9,.0f}
    """
    
    ax4.text(0.1, 0.5, metrics_text, fontsize=12, family='monospace',
             verticalalignment='center')
    
    plt.tight_layout()
    plt.show()
    return fig


fig = plot_clean_performance(simulator)

# ============================================================
# STEP 6: VERIFICATION CHECKS
# ============================================================

print("\n" + "="*60)
print("VERIFICATION CHECKS")
print("="*60)

# Check 1: Roll events
roll_trades = [t for t in simulator.trades if 'ROLL' in t['type']]
print(f"\nRoll events: {len(roll_trades)}")
print("First 10 roll dates:")
for i, trade in enumerate(roll_trades[:10]):
    print(f"  Roll {i+1}: {trade['date']} ({trade['type']})")

# Check 2: Position consistency
print("\nPosition consistency check:")
print(f"  Long calendars: {simulator.long_calendars}")
print(f"  Short calendars: {simulator.short_calendars}")
if simulator.short_calendars > 0 or simulator.long_calendars > 0:
    print(f"  ⚠ Warning: Open positions at end of backtest")

# Check 3: P&L breakdown
print("\nP&L breakdown:")
print(f"  Starting cash: $100,000.00")
print(f"  Final cash: ${simulator.cash:,.2f}")
open_position_value = simulator.daily_portfolio_value[-1] - simulator.cash
print(f"  Open position value: ${open_position_value:,.2f}")
print(f"  Total portfolio: ${simulator.daily_portfolio_value[-1]:,.2f}")
print(f"  P&L = ${simulator.daily_portfolio_value[-1] - simulator.initial_cash:,.2f}")

# Check 4: Trade reduction from holding period
signal_changes = sum(1 for i in range(1, len(df_real)) 
                     if df_real.iloc[i]['signal'] != df_real.iloc[i-1]['signal'])
actual_trades = len([t for t in simulator.trades if 'ROLL' not in t['type']])
print(f"\nTurnover analysis:")
print(f"  Signal changes: {signal_changes}")
print(f"  Actual trades (excl rolls): {actual_trades}")
print(f"  Rolls: {len(roll_trades)}")

print("="*60)
print("\n✓ Simulation v2 complete!")
```

```python
"""
ABLATION TEST: Run 4 configs to isolate which improvement helps/hurts.
Paste this as a new cell after your simulation cells.
"""

def run_ablation(df_data, config, label):
    """Run full simulation with a given config and return metrics."""
    
    # Generate signals
    df = df_data.copy()
    df['signal'] = df.apply(
        lambda row: generate_signal(row, config) 
        if pd.notna(row.get('slope_forecast')) 
        else {'long_calendars': 0, 'short_calendars': 0},
        axis=1
    )
    
    # Apply holding period
    df = apply_holding_period(df, config.min_hold_days)
    
    # Run simulation
    sim = CalendarSpreadSimulator(initial_cash=100000)
    
    for idx, row in df.iterrows():
        # Expiration handling: use OLD logic if roll_days_before_expiry == 0
        if config.roll_days_before_expiry == 0:
            # Original expiration logic (settle at intrinsic + close 60d leg)
            if (idx + 1) % config.expiry_cycle_days == 0:
                row_data = df.iloc[idx]
                spx_price = row_data['spx_price']
                strike = row_data['strike']
                intrinsic_value_30 = max(spx_price - strike, 0)
                
                if sim.long_calendars > 0:
                    settlement = -sim.long_calendars * intrinsic_value_30
                    closing_value = sim.long_calendars * row_data['option_60dte_mid']
                    sim.cash += settlement + closing_value
                    sim.long_calendars = 0
                
                if sim.short_calendars > 0:
                    settlement = sim.short_calendars * intrinsic_value_30
                    closing_cost = -sim.short_calendars * row_data['option_60dte_mid']
                    sim.cash += settlement + closing_cost
                    sim.short_calendars = 0
        else:
            # New roll logic
            handle_expiration(sim, idx, df, config)
        
        execute_trades(sim, row, use_costs=True)
        
        long_cal_mid = row['option_60dte_mid'] - row['option_30dte_mid']
        short_cal_mid = row['option_30dte_mid'] - row['option_60dte_mid']
        portfolio_value = sim.get_portfolio_value(long_cal_mid, short_cal_mid)
        
        if idx == 0:
            daily_pnl = 0
        else:
            daily_pnl = portfolio_value - sim.daily_portfolio_value[-1]
        
        sim.daily_pnl.append(daily_pnl)
        sim.daily_portfolio_value.append(portfolio_value)
    
    # Calculate metrics
    daily_returns = np.diff(sim.daily_portfolio_value) / sim.daily_portfolio_value[:-1]
    sharpe = (np.mean(daily_returns) / np.std(daily_returns)) * np.sqrt(252) if np.std(daily_returns) > 0 else 0
    total_return_pct = (sim.daily_portfolio_value[-1] / sim.initial_cash - 1) * 100
    portfolio_values = np.array(sim.daily_portfolio_value)
    max_dd = ((portfolio_values - np.maximum.accumulate(portfolio_values)) / np.maximum.accumulate(portfolio_values) * 100).min()
    
    return {
        'label': label,
        'sharpe': sharpe,
        'total_return_pct': total_return_pct,
        'max_dd': max_dd,
        'num_trades': len(sim.trades),
        'txn_costs': sim.transaction_costs,
        'portfolio_values': sim.daily_portfolio_value
    }


# ── Define test configs ──────────────────────────────────────
configs = [
    (RuleConfig(base_aggressiveness=5, dynamic_agg=False, min_hold_days=1, 
                roll_days_before_expiry=0, expiry_cycle_days=30),
     "Baseline (v1)"),
    
    (RuleConfig(base_aggressiveness=5, dynamic_agg=True, min_hold_days=1, 
                roll_days_before_expiry=0, expiry_cycle_days=30),
     "+ Dynamic Agg only"),
    
    (RuleConfig(base_aggressiveness=5, dynamic_agg=False, min_hold_days=3, 
                roll_days_before_expiry=0, expiry_cycle_days=30),
     "+ Holding Period only"),
    
    (RuleConfig(base_aggressiveness=5, dynamic_agg=False, min_hold_days=1, 
                roll_days_before_expiry=3, expiry_cycle_days=30),
     "+ Smart Roll only"),
    
    (RuleConfig(base_aggressiveness=5, dynamic_agg=True, min_hold_days=3, 
                roll_days_before_expiry=3, expiry_cycle_days=30),
     "All combined (v2)"),
]

# ── Run all tests ────────────────────────────────────────────
results = []
for cfg, label in configs:
    print(f"Running: {label}...")
    res = run_ablation(df_real, cfg, label)
    results.append(res)

# ── Print comparison table ───────────────────────────────────
print("\n" + "="*85)
print(f"{'Config':<25} {'Sharpe':>8} {'Return%':>10} {'MaxDD%':>10} {'Trades':>8} {'TxnCost':>12}")
print("="*85)
for r in results:
    print(f"{r['label']:<25} {r['sharpe']:>8.2f} {r['total_return_pct']:>9.1f}% {r['max_dd']:>9.1f}% {r['num_trades']:>8} ${r['txn_costs']:>10,.0f}")
print("="*85)

# ── Plot equity curves ───────────────────────────────────────
plt.figure(figsize=(14, 6))
for r in results:
    returns = (np.array(r['portfolio_values']) / 100000 - 1) * 100
    plt.plot(returns, linewidth=2, label=r['label'], alpha=0.8)

plt.axhline(y=0, color='black', linestyle='--', linewidth=1, alpha=0.5)
plt.title('Ablation Test: Equity Curves by Configuration', fontsize=14, fontweight='bold')
plt.ylabel('Return (%)')
plt.xlabel('Trading Days')
plt.legend(loc='upper left', fontsize=10)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
